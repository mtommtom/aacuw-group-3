```python
import pandas as pd

df = pd.read_csv('../cleaned datatsets/bitcoin.csv', delimiter=';')
df = df[['timestamp', 'open', 'high', 'low', 'close', 'volume', 'marketCap', 'circulatingSupply']]
df['timestamp'] = pd.to_datetime(df['timestamp'])
```

I am creating a year and month column because I am bad with working on timeseries so this is easier

```python
df['year'] = df['timestamp'].dt.year
df['month'] = df['timestamp'].dt.month

df = df.set_index('timestamp')
df.head()
```

Print market cap over time

```python
import seaborn as sns
import matplotlib.pylab as plt
```

```python
# %config InlineBackend.figure_format = 'retina'
# plt.style.use('https://github.com/dhaitz/matplotlib-stylesheets/raw/master/pitayasmoothie-dark.mplstyle')

import mplcyberpunk
plt.style.use('cyberpunk')
```

```python
sns.set_style()
ax = sns.lineplot(df, x=df.index, y='marketCap')
ax.set_xlabel("time")
ax.set_ylabel("market cap")
ax.set_title("Bitcoin market cap over time")
mplcyberpunk.add_glow_effects()
```
